```python
"""
Extracts the maximum recorded wind speed for each TC with at least Category 1 intensity 
and adds to the hurricaneOne.csv dataset.

Author: Suchit Basineni
Date: 7/9/24
"""

import pandas as pd

df = pd.read_csv('HurricaneData/ibtracs.NA.list.v04r00.csv')

df['USA_WIND'] = pd.to_numeric(df['USA_WIND'], errors='coerce').fillna(-1)

filtered_df = df[df['USA_SSHS'] >= 1]

highest_windspeed_per_sid = filtered_df.loc[
    filtered_df.groupby('SID')['USA_WIND'].idxmax()
]

print(highest_windspeed_per_sid)

highest_windspeed_per_sid.to_csv('HurricaneData/hurricaneOne.csv', index=False)
```
